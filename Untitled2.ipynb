```python
import numpy as np

# 1. 2 x 2 array
a = np.array([[1, 2], [3, 4]])
print("1. 2x2 array:\n", a, "\n")

# 2. Random 3 x 4 array
r = np.random.rand(3, 4)
print("2. Random 3x4 array:\n", r, "\n")

# 3. Ones and zeros
print("3. Ones:\n", np.ones((2, 3)), "\n")
print("3. Zeros:\n", np.zeros((2, 3)), "\n")

# 4. Random 6 x 3 array and its transpose
b = np.random.rand(6, 3)
print("4. Original (6x3):\n", b, "\n")
print("4. Transposed (3x6):\n", b.T, "\n")

# 5. Dot product of (n x p) and (p x m), with n=4, p=3, m=5
n, p, m = 4, 3, 5
A = np.random.rand(n, p)
B = np.random.rand(p, m)
C = np.dot(A, B)
print("5. Dot product result (4x5):\n", C)
```
